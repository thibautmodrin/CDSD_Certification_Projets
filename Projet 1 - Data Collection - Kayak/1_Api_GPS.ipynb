```python
cities=["Mont Saint Michel",
"St Malo",
"Bayeux",
"Le Havre",
"Rouen",
"Paris",
"Amiens",
"Lille",
"Strasbourg",
"Chateau du Haut Koenigsbourg",
"Colmar",
"Eguisheim",
"Besancon",
"Dijon",
"Annecy",
"Grenoble",
"Lyon",
# "Gorges du Verdon",
"Bormes les Mimosas",
"Cassis",
"Marseille",
"Aix en Provence",
"Avignon",
"Uzes",
"Nimes",
"Aigues Mortes",
"Saintes Maries de la mer",
"Collioure",
"Carcassonne",
# "Ariege",
"Toulouse",
"Montauban",
"Biarritz",
"Bayonne",
"La Rochelle"]
```

```python
import requests
import json


def get_lat_lon(city):
    url = "https://nominatim.openstreetmap.org/search?"

    headers = {
        'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_10_1) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/3'
    }
    # Paramètres de la requête
    params = {
        'city': city,
        'format':'json'
    }

    # Envoi de la requête
    response = requests.get(url, params=params,headers=headers)

    data=response.json()
    lat=data[0]['lat']
    lon=data[0]['lon']
    return lat,lon
```

```python

cities_lat_long_dict={}
for city in cities:
    try:
        lat,lon=get_lat_lon(city)
        cities_lat_long_dict[city]=[lat,lon]
    except Exception as e:
        print(f"Error retrieving data for {city}: {e}")
        cities_lat_long_dict[city]=[None,None]



```

```python
import pandas
df=pandas.DataFrame.from_dict(cities_lat_long_dict,orient='index',columns=['lat','lon'])
df['lat']=df['lat'].astype(float)
df['lon']=df['lon'].astype(float)
```
