```python
import json
import os
import pandas as pd
import boto3
from dotenv import load_dotenv
```

```python

load_dotenv()

aws_access_key_id = os.getenv('AWS_ACCESS_KEY_ID')
aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY')
s3 = boto3.resource('s3',aws_access_key_id=aws_access_key_id,aws_secret_access_key=aws_secret_access_key,region_name='eu-west-3')
bucket=s3.Bucket('tmopenlabbucket')
```

```python
bucket=s3.create_bucket(Bucket='tmopenlabbucket',CreateBucketConfiguration={'LocationConstraint':'eu-west-3'})
```

```python
bucket.upload_file(Filename='cities_lat_long_ccm.csv',Key='City_Meteo_Rank.csv', ExtraArgs={'ACL':'public-read'})
```

```python
s3_client = boto3.client('s3')

# Le nom du bucket à modifier
bucket_name = 'tmopenlabbucket'

# 1. Modifier Object Ownership
s3_client.put_bucket_ownership_controls(
    Bucket=bucket_name,
    OwnershipControls={
        'Rules': [
            {
                'ObjectOwnership': 'BucketOwnerPreferred'  # ou 'BucketOwnerEnforced', 'ObjectWriter'
            }
        ]
    }
)



# 2. Modifier Block Public Access
response_public_access = s3_client.put_public_access_block(
    Bucket=bucket_name,
    PublicAccessBlockConfiguration={
        'BlockPublicAcls': False,      # Bloquer les ACLs publiques
        'IgnorePublicAcls': False,     # Ignorer toutes les ACLs publiques
        'BlockPublicPolicy': False,    # Empêcher les politiques publiques
        'RestrictPublicBuckets': False # Restreindre les buckets publics
    }
)
```
