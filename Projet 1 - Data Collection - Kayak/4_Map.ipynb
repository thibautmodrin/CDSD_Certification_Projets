```python
import requests
from io import StringIO
import pandas as pd
import plotly.express as px


response=requests.get('https://tmopenlabbucket.s3.eu-west-3.amazonaws.com/City_Meteo_Rank.csv')
df = pd.read_csv(StringIO(response.text))
df.drop(columns=['Unnamed: 0'], inplace=True)
df
```

```python
fig=px.scatter_mapbox(df,lat='lat',lon='lon',color='CCM',mapbox_style='open-street-map',zoom=5,size='CCM',color_continuous_scale=px.colors.sequential.speed,height=1000
                      )

fig.show()
```
