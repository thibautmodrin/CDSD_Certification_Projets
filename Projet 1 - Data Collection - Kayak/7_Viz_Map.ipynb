```python
##### Get dataframe by city name with hostels information name, score,description, latitude, longitude

def get_city_info(city_name):
    city_info = df[df['city'].str.lower() == city_name.lower()]
    if not city_info.empty:
        return city_info[['city','url', 'CCM', 'description', 'latitude', 'longitude']].reset_index(drop=True)
    else:
        return pd.DataFrame(columns=['city','url', 'CCM', 'description', 'latitude', 'longitude']).reset_index()
    
```

```python
import math

def calculate_zoom_level(lat_min, lat_max, lon_min, lon_max, map_width=800, map_height=600):
    # Calcul approximatif du niveau de zoom basé sur les limites
    lat_diff = lat_max - lat_min
    lon_diff = lon_max - lon_min
    
    # Facteurs d'ajustement
    lat_zoom = math.log2(360 / lat_diff) if lat_diff > 0 else 10
    lon_zoom = math.log2(360 / lon_diff) if lon_diff > 0 else 10
    
    return min(lat_zoom, lon_zoom)  # -1 pour avoir un peu de marge

```

```python
#### Map best hostel for each city with ScatterMap and Dash

import dash
from dash import dcc, html
from dash.dependencies import Input, Output
from dash import Dash
import plotly.express as px
from plotly import graph_objects as go
from IPython.display import HTML
import numpy as np  # <-- nécessaire pour customdata

# enlever warning depreciation
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)

HTML('<style>.jp-OutputArea-output { overflow: visible !important; }</style>')

DEFAULT_CITY = "Paris"

app = Dash(__name__)
app.config.suppress_callback_exceptions = True  # sécurise si le layout évolue

app.layout = html.Div([
    dcc.Graph(
        id='city-map',
        figure=fig,
        config={
            'scrollZoom': True,       # Zoom avec molette
            'doubleClick': 'reset'    # Double-clic pour reset
        }
    ),
    dcc.Graph(
        id='hostel-map',
        config={
            'scrollZoom': True,      # Zoom avec molette
            'doubleClick': 'reset',  # Double-clic pour reset
            'displayModeBar': True,  # Afficher la barre d'outils
        }
    ),
    # Store caché: sert d'Output au callback JS pour éviter les conflits d'outputs
    dcc.Store(id='open-url-trigger')
], style={
    'display': 'flex',
    'flex-direction': 'row',
    'justify-content': 'justify',
    'align-items': 'flex-start'
})

@app.callback(
    Output('hostel-map', 'figure'),
    Input('city-map', 'clickData'),
    prevent_initial_call=False  # on veut un passage au chargement
)
def update_hostel_map(clickData):
    # 1) Choisir la ville : Paris par défaut si pas de clic
    if clickData and clickData.get('points'):
        pt = clickData['points'][0]
        city_name = pt.get('text') or pt.get('customdata') or DEFAULT_CITY
    else:
        city_name = DEFAULT_CITY

    # 2) Infos de la ville
    city_info = get_city_info(city_name)
    if city_info.empty:
        return go.Figure()  # <-- un seul Output => on ne renvoie qu'une figure

    # 3) Filtrer les hostels de la ville
    city_df = df[df['city'].str.lower() == city_name.lower()]
    if city_df.empty:
        return go.Figure()  # <-- idem

    # 4) Zoom auto
    zoom_level = calculate_zoom_level(
        city_df['latitude'].min(),
        city_df['latitude'].max(),
        city_df['longitude'].min(),
        city_df['longitude'].max()
    )

    # 5) Figure des hostels
    #    On met l'URL Booking dans customdata (non affichée) pour utilisation au clic
    #    Assure-toi que df contient une colonne 'url' avec des URLs http(s) complètes
    if 'url' not in city_df.columns:
        city_df['url'] = "about:blank"

    hostel_fig = go.Figure(go.Scattermapbox(
        lat=city_df['latitude'],
        lon=city_df['longitude'],
        mode='markers+text',
        # On garde ton hover simple, sans URL
        hoverinfo='text',
        hovertext=city_df.apply(
            lambda row: f"Hostel: {row['name']}<br>Score: {row['score']}",
            axis=1
        ),
        # URL transportée de façon invisible pour le clientside_callback
        customdata=np.stack([city_df['url'].astype(str)], axis=-1),
        marker=dict(
            size=(city_df['score'] * 3).clip(lower=6),  # tailles min pour lisibilité
            color=city_df['score'],
            showscale=True,
            colorbar=dict(title='Hostel Score'),
            colorscale=[
                [0.0, '#d32f2f'],    # Rouge foncé (mauvais)
                [0.2, '#f57c00'],    # Orange (moyen-)
                [0.4, '#fbc02d'],    # Jaune (moyen)
                [0.6, '#689f38'],    # Vert clair (bien)
                [0.8, '#388e3c'],    # Vert (très bien)
                [1.0, '#1b5e20']     # Vert foncé (excellent)
            ],
        ),
    ))

    # 6) Layout de la carte
    hostel_fig.update_layout(
        title=f'Best Hostels in {city_name}',
        mapbox=dict(
            style='open-street-map',  # ou 'carto-positron' si tu préfères
            center=dict(
                lat=float(city_info['latitude'].mean()),
                lon=float(city_info['longitude'].mean())
            ),
            zoom=float(zoom_level)
        ),
        height=800,
        width=800,
        margin=dict(l=0, r=0, t=50, b=0)
    )

    return hostel_fig

# --- Callback clientside (JS) : ouvre l'URL au clic d'un cercle (aucun lien affiché dans le hover dans le navigateur uniquement) ---
app.clientside_callback(
    """
    function(clickData){
        if (!clickData || !clickData.points || !clickData.points.length){
            return window.dash_clientside.no_update;
        }
        try {
            const pt = clickData.points[0] || null;
            const cd = pt && pt.customdata;
            let url = null;

            if (Array.isArray(cd) && cd.length > 0){
                url = cd[0];
            } else if (typeof cd === 'string'){
                url = cd;
            } else if (cd && typeof cd === 'object' && (0 in cd)){
                url = cd[0];
            }

            if (typeof url === 'string' && /^https?:\\/\\//.test(url)){
                window.open(url, '_blank');  // ouvre l'annonce Booking
            }
        } catch(e) {
            // console.warn(e);
        }
        return window.dash_clientside.no_update;
    }
    """,
    Output('open-url-trigger', 'data'),   # sortie vers le Store caché
    Input('hostel-map', 'clickData'),
    prevent_initial_call=True
)

if __name__ == '__main__':
    app.run(
        debug=True,
        port=8050,
        use_reloader=True,       # évite les callbacks orphelins en Notebook
        dev_tools_open_browser=True,
        dev_tools_hot_reload=True,
        jupyter_height='1200px',
        
)
```
